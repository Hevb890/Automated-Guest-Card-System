```python
import pandas as pd
from PIL import Image, ImageDraw, ImageFont
import os
```

```python
# Configurations
PSD_PATH = "Base.jpg"
EXCEL_PATH = "Names.xlsx"
FONT_PATH = "Cormorant-Bold.otf"
OUTPUT_FOLDER = "./Invitees"
FONT_SIZE = 60
TEXT_COLOR = (0,0,0)
```

```python
box_coords = [100, -100, 1320, 1400]
```

```python
if not os.path.exists(OUTPUT_FOLDER):
    os.makedirs(OUTPUT_FOLDER)
```

```python

df = pd.read_excel(EXCEL_PATH)
font = ImageFont.truetype(FONT_PATH, FONT_SIZE)
```

```python
df
```

```python
for index, row in df.iterrows():
    name = str(row["Names"]).strip()
    img = Image.open(PSD_PATH)
    
    draw = ImageDraw.Draw(img)
   
    bbox = draw.textbbox((0, 0), name, font=font)
    text_width = bbox[2] - bbox[0]
    text_height = bbox[3] - bbox[1]
    

    # x = (box_coords[2] - box_coords[0] - text_width) / 2 + box_coords[0]
    x = box_coords[2] - text_width
    # Y = (BoxHeight - TextHeight) / 2 + TopMargin
    y = (box_coords[3] - box_coords[1] - text_height) / 2 + box_coords[1]
    
    # Draw the text
    draw.text((x, y), name, fill=TEXT_COLOR, font=font)
    
    # Save as JPEG for easy WhatsApp sharing
    file_name = f"{name.replace(' ', '_')}_invite.jpg"
    img.save(os.path.join(OUTPUT_FOLDER, file_name), "JPEG", quality=95)
    print(f"Generated: {file_name}")
```
